## Task 1: Ingest IPL Player Stats & Detect Missing Values

**Problem Statement:**  
*Download a sample CSV file of IPL player stats (include columns like player name, runs, matches, and some missing values). Load it into a pandas DataFrame and use `isnull()` and `notnull()` to print how many missing values are present in each column.*


```python
import pandas as pd
import numpy as np

# Load IPL player stats dataset
df_ipl = pd.read_csv('ipl_player_stats.csv')
print('Dataset Preview:')
print(df_ipl.head(5))

# Method 1: Count missing values using .isnull().sum()
missing_isnull = df_ipl.isnull().sum()

# Method 2: Calculate missing values using .notnull()
# Total rows minus count of valid values gives missing count
missing_notnull = len(df_ipl) - df_ipl.notnull().sum()

print('\nMissing values count per column (using isnull):')
print(missing_isnull)

print('\nMissing values count per column (verified using notnull):')
print(missing_notnull)
```

## Task 2: Remove Missing Rows with `dropna(axis=0, how='any')`

**Problem Statement:**  
*Using the loaded IPL player stats DataFrame, apply `dropna(axis=0, how='any')` to remove all rows with any missing data and display the shape of the DataFrame before and after dropping.*


```python
# Inspect shape before dropping
shape_before = df_ipl.shape

# Apply dropna to remove any row containing at least one missing value
df_dropped = df_ipl.dropna(axis=0, how='any')
shape_after = df_dropped.shape

print(f'DataFrame Shape Before dropna: {shape_before}')
print(f'DataFrame Shape After dropna:  {shape_after}')
print(f'Rows removed due to missing values: {shape_before[0] - shape_after[0]}')

print('\nCleaned DataFrame (all complete rows):')
df_dropped
```

## Task 3: Mean Imputation on 'runs' Column with `fillna()`

**Problem Statement:**  
*For the 'runs' column in your IPL player stats DataFrame, use `fillna()` to replace missing values with the mean of the column. Print the updated column to verify the changes.*


```python
# Compute the mean of the 'runs' column (excluding nulls)
runs_mean = df_ipl['runs'].mean()
print(f'Calculated Mean Runs: {runs_mean:.2f}')

# Compare original vs imputed runs column
df_ipl_imputed = df_ipl.copy()
df_ipl_imputed['runs'] = df_ipl_imputed['runs'].fillna(runs_mean)

# Display player name with original and updated runs to verify imputation
verification_df = pd.DataFrame({
    'player_name': df_ipl['player_name'],
    'runs_original': df_ipl['runs'],
    'runs_imputed': df_ipl_imputed['runs'].round(2)
})

print('\nVerification Table (Missing runs imputed with mean):')
print(verification_df)

print(f'\nMissing values in updated runs column: {df_ipl_imputed["runs"].isnull().sum()}')
```

## Task 4: Sequential Fill (Forward Fill + Backward Fill)

**Problem Statement:**  
*Simulate a Zomato-style restaurant ratings dataset with some missing ratings. Use forward fill (`method='ffill'` or `.ffill()`) to fill missing values in the ratings column, then use backward fill (`method='bfill'` or `.bfill()`) for any remaining missing values. Show the before and after results.*


```python
# Load simulated Zomato restaurant ratings dataset
df_zomato = pd.read_csv('zomato_ratings.csv')

# Create copy for sequential imputation
df_ratings = df_zomato.copy()

# Step 1: Forward fill propagates the last valid rating forward
df_ratings['rating_ffill'] = df_ratings['rating'].ffill()

# Step 2: Backward fill handles leading NaNs that ffill cannot reach
df_ratings['rating_final'] = df_ratings['rating_ffill'].bfill()

print('Comparison Table: Before vs After Sequential Imputation:')
comparison = df_ratings[['restaurant_name', 'rating', 'rating_ffill', 'rating_final']]
comparison
```

## Task 5: Missingness Mechanism Analysis (MCAR, MAR, MNAR)

**Problem Statement:**  
*Pick any one column with missing values from your dataset and explain which missingness mechanism (MCAR, MAR, or MNAR) is most likely and why, using 2-3 sentences.*  
*Hint: Think about whether the missing data is random or related to other factors in the dataset.*


### 🔬 Missingness Mechanism Evaluation

#### **Selected Column:** `wickets` (from `ipl_player_stats.csv`)

> **Most Likely Mechanism:** **MAR (Missing at Random)**
> 
> **Explanation:**  
> In the IPL dataset, missing values in the `wickets` column are not occurring completely at random; rather, they are systematically related to the player's primary role in the team. Specialist batsmen and wicketkeepers (such as Virat Kohli, Heinrich Klaasen, or MS Dhoni) rarely or never bowl in a match, meaning their bowling stats are unrecorded or absent conditioned directly on their observed batting specialty. Because the probability of missingness can be fully explained by another observed variable (`player role` / `runs`), it represents **Missing at Random (MAR)**.

---

#### *Alternative Perspective on Zomato `rating` Column:*
> In restaurant review datasets, missing ratings often exemplify **MNAR (Missing Not at Random)**. Diners with neutral or average experiences are significantly less inclined to leave any rating on the app compared to customers with exceptionally positive or negative experiences. Thus, the missingness depends directly on the unobserved customer sentiment itself.

## 🎯 Key Takeaways & Best Practices
- **Detection**: Always begin EDA with `df.isnull().sum()` or visual heatmaps to assess missingness proportion per feature.
- **Deletion vs. Imputation**:
  - `dropna(how='any')` can cause drastic data loss (here reducing rows from 10 to 5, a 50% drop!). Use it only when missing values are scarce (< 2-3%).
  - Imputation preserves sample size. Use **Mean** for symmetric distributions, **Median** for skewed data/outliers, and **Mode** for categorical data.
- **Sequential Imputation**: For time-series, order logs, or sequential records, pairing `ffill()` followed by `bfill()` ensures all gaps (including leading missing values) are filled.
- **Understanding Mechanisms**: Recognizing whether data is **MCAR**, **MAR**, or **MNAR** prevents introducing bias into downstream machine learning models.
