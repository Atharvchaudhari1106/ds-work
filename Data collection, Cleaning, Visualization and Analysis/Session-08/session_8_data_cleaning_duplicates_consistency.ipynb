## Task 1: Find Duplicate Zomato Orders with `duplicated()`

**Problem Statement:**  
*Download a small sample of your recent Zomato order history as a CSV (or create a mock CSV with restaurant names and order dates), then use pandas' `duplicated()` function to find and print any duplicate orders based on restaurant name and date.*


```python
import pandas as pd

# Load Zomato order history dataset
df_zomato = pd.read_csv('zomato_order_history.csv')
print('Zomato Orders Dataset Preview:')
print(df_zomato.head(5))

# Identify duplicate orders based on restaurant name and order date
# Using keep=False marks all occurrences of duplicates for complete audit
duplicate_mask_all = df_zomato.duplicated(subset=['restaurant_name', 'order_date'], keep=False)
duplicate_orders_all = df_zomato[duplicate_mask_all].sort_values(by=['restaurant_name', 'order_date'])

print(f'\nTotal orders flagged in duplicate clusters: {len(duplicate_orders_all)}')
print('\nAll duplicate order occurrences (matching restaurant & date):')
print(duplicate_orders_all)

# Using keep='first' marks only the redundant duplicate rows (excluding original)
duplicate_mask_subsequent = df_zomato.duplicated(subset=['restaurant_name', 'order_date'], keep='first')
duplicate_orders_subsequent = df_zomato[duplicate_mask_subsequent]

print(f'\nNumber of redundant rows to drop: {len(duplicate_orders_subsequent)}')
print('Redundant rows:')
duplicate_orders_subsequent
```

## Task 2: Identify Repeated Reviews with `value_counts()`

**Problem Statement:**  
*Given a list of Flipkart product reviews with some duplicate entries, use `value_counts()` in pandas to identify which review texts are repeated most often and display the top 3 most common duplicate reviews.*


```python
# Load Flipkart product reviews dataset
df_reviews = pd.read_csv('flipkart_reviews.csv')

print(f'Total reviews in dataset: {len(df_reviews)}')

# Count frequency of each unique review text
review_counts = df_reviews['review_text'].value_counts()

# Extract top 3 most frequently occurring duplicate reviews
top_3_duplicate_reviews = review_counts.head(3)

print('Top 3 Most Common Duplicate Reviews:')
for rank, (text, count) in enumerate(top_3_duplicate_reviews.items(), 1):
    print(f'{rank}. "{text}" -> Repeated {count} times')

print('\nDetailed Series Output:')
top_3_duplicate_reviews
```

## Task 3: Exact Duplicate Removal with `drop_duplicates()`

**Problem Statement:**  
*Create a DataFrame with mock data for Spotify playlists, including playlist names and creator usernames, where some rows are exact duplicates. Use `drop_duplicates()` to remove duplicate playlists and print the cleaned DataFrame.*


```python
# Load Spotify playlists dataset with exact duplicate rows
df_spotify = pd.read_csv('spotify_playlists.csv')

print(f'Initial DataFrame Shape: {df_spotify.shape}')
print('Initial Dataset (with exact duplicate rows):')
print(df_spotify)

# Remove exact duplicate rows and reset index
df_spotify_cleaned = df_spotify.drop_duplicates().reset_index(drop=True)

print(f'\nCleaned DataFrame Shape: {df_spotify_cleaned.shape}')
print(f'Duplicates removed: {df_spotify.shape[0] - df_spotify_cleaned.shape[0]} rows')

print('\nCleaned Spotify Playlists DataFrame:')
df_spotify_cleaned
```

## Task 4: Standardize Typos in Categorical Data with `replace()`

**Problem Statement:**  
*Suppose you have a DataFrame of Instagram usernames where some entries have typos (like `'insta_queen'`, `'insta queen'`, `'instaqueen'`). Use the `replace()` function to standardize all these variants to `'instaqueen'`.*


```python
# Load Instagram users dataset with username variants
df_insta = pd.read_csv('instagram_users.csv')

print('Original Instagram Usernames:')
print(df_insta[['user_id', 'username']])

# Define dictionary mapping typo variants to the canonical username
typo_mapping = {
    'insta_queen': 'instaqueen',
    'insta queen': 'instaqueen'
}

# Standardize username variants
df_insta['username_clean'] = df_insta['username'].replace(typo_mapping)

print('\nStandardized Instagram Usernames:')
comparison = df_insta[['user_id', 'username', 'username_clean', 'followers_count']]
print(comparison)

print('\nUnique counts before vs after cleaning:')
print(f'Before: {df_insta["username"].nunique()} unique usernames')
print(f'After:  {df_insta["username_clean"].nunique()} unique usernames')
```

## Task 5: Normalize Inconsistent Payment Status to Binary Flags

**Problem Statement:**  
*You have a DataFrame column for payment status from a Paytm-like app with mixed values: `'Yes'`, `'yes'`, `'Y'`, `'No'`, `'no'`, `'N'`, and some with extra spaces. Write code to unify this column so all paid statuses become 1 and all unpaid statuses become 0, trimming whitespace and fixing capitalization where needed.*  
*Hint: Use `str.strip()`, `str.lower()`, and map/replace methods in pandas.*


```python
# Load Paytm payment status dataset
df_paytm = pd.read_csv('paytm_payment_status.csv')

print('Original Raw Payment Status Column (with spaces & mixed cases):')
for val in df_paytm['payment_status']:
    print(f'Raw representation: {repr(val)}')

# Step 1: Strip leading/trailing whitespaces and convert to lowercase
cleaned_status = df_paytm['payment_status'].str.strip().str.lower()

# Step 2: Define mapping to binary indicator flags (1 = Paid, 0 = Unpaid)
status_binary_map = {
    'yes': 1,
    'y': 1,
    'no': 0,
    'n': 0
}

# Step 3: Map cleaned strings to integer binary flags
df_paytm['is_paid'] = cleaned_status.map(status_binary_map)

print('\nUnified Payment Status Table:')
result_table = df_paytm[['txn_id', 'customer_id', 'payment_status', 'is_paid']]
print(result_table)
print(f'\nTarget column data type: {df_paytm["is_paid"].dtype}')
```

## 🎯 Key Takeaways & Best Practices
- **Duplication Diagnostics**: Use `duplicated(keep=False)` to inspect entire duplicate clusters before dropping; use `duplicated(keep='first')` to single out redundant occurrences.
- **Frequency Auditing**: `.value_counts()` serves as an essential preliminary check to catch spam bots, repeated reviews, or duplicate default entries.
- **Idempotent De-duplication**: `drop_duplicates()` eliminates exact row matches cleanly. When keys dictate uniqueness (e.g. `order_id` or `user_id`), specify `subset=['key_column']`.
- **Systematic Typo Remediation**: When handling small controlled vocabularies, `.replace(dict_mapping)` provides deterministic, safe corrections. For large-scale fuzzy typos, consider phonetic algorithms or Levenshtein distance libraries (`thefuzz`).
- **Data Hygiene Pipeline**: String data requires standard sequential hygiene: `.str.strip()` $\rightarrow$ `.str.lower()` $\rightarrow$ `.map(binary_dict)` to avoid silent parsing mismatches.
