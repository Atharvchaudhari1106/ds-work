## Task 1: Convert String Delivery Timestamps to Datetime

**Problem Statement:**  
*Given a list of delivery timestamps as strings (e.g., `['2024-06-01 14:30', '2024-06-02 09:15', '2024-06-03 20:45']`), use pandas and `pd.to_datetime()` to convert them into datetime objects and print the result.*


```python
import pandas as pd
import numpy as np

# Input list of string delivery timestamps
delivery_timestamps = ['2024-06-01 14:30', '2024-06-02 09:15', '2024-06-03 20:45']

# Convert string list to pandas DatetimeIndex
datetime_objects = pd.to_datetime(delivery_timestamps)

print('Converted Datetime Objects:')
print(datetime_objects)
print(f'\nResult Dtype: {datetime_objects.dtype}')
print(f'Underlying object type: {type(datetime_objects)}')
```

## Task 2: Extract Year, Month, and Weekday from Order Dates

**Problem Statement:**  
*Load a CSV file containing order dates from a Flipkart-style order history (column: `'order_date'`, format: `'YYYY-MM-DD HH:MM:SS'`). Extract the year, month, and weekday for each order and add them as new columns in the DataFrame.*


```python
# Load Flipkart order history dataset
df_flipkart = pd.read_csv('flipkart_orders.csv')

# Convert 'order_date' column to datetime type
df_flipkart['order_date'] = pd.to_datetime(df_flipkart['order_date'], format='%Y-%m-%d %H:%M:%S')

# Extract datetime components using the .dt accessor
df_flipkart['year'] = df_flipkart['order_date'].dt.year
df_flipkart['month'] = df_flipkart['order_date'].dt.month
df_flipkart['weekday'] = df_flipkart['order_date'].dt.day_name()

print('Flipkart Orders with Extracted Temporal Components (First 5 Rows):')
preview = df_flipkart[['order_id', 'customer_id', 'order_date', 'year', 'month', 'weekday', 'amount']]
preview.head(5)
```

## Task 3: Datetime Indexing & Weekly Order Resampling

**Problem Statement:**  
*Set the `'order_date'` column as the index of your DataFrame and use resampling to calculate the total number of orders placed each week.*  
*Hint: Use `df.resample('W').size()` after setting the datetime index.*


```python
# Set 'order_date' as the DataFrame index
df_indexed = df_flipkart.set_index('order_date')

# Resample data on a weekly basis ('W') and compute order counts
weekly_order_counts = df_indexed.resample('W').size()

print('Weekly Total Orders Placed (Resampled):')
print(weekly_order_counts)

print(f'\nTotal orders aggregated across all weeks: {weekly_order_counts.sum()}')
```

## Task 4: Timezone Localization & Conversion (UTC to Asia/Kolkata)

**Problem Statement:**  
*Suppose you have a DataFrame of Instagram posts with a `'posted_at'` column in UTC. Convert these timestamps to `'Asia/Kolkata'` timezone and display the first 5 converted times.*


```python
# Load Instagram posts dataset
df_insta = pd.read_csv('instagram_posts.csv')

# Step 1: Parse to datetime and localize as UTC timezone-aware
df_insta['posted_at_utc'] = pd.to_datetime(df_insta['posted_at']).dt.tz_localize('UTC')

# Step 2: Convert timezone from UTC to 'Asia/Kolkata' (IST = UTC + 5:30)
df_insta['posted_at_ist'] = df_insta['posted_at_utc'].dt.tz_convert('Asia/Kolkata')

print('Instagram Posts - UTC vs IST Timezone Comparison (First 5):')
tz_comparison = df_insta[['post_id', 'username', 'posted_at_utc', 'posted_at_ist']]
tz_comparison.head(5)
```

## Task 5: Feature Engineering - `is_weekend` Flag

**Problem Statement:**  
*Create a new feature called `'is_weekend'` in your DataFrame that marks `True` if an order was placed on Saturday or Sunday, and `False` otherwise.*  
*Constraint: Do not use any external libraries except pandas and numpy.*


```python
# In pandas, dt.dayofweek represents: Monday=0, Tuesday=1, ..., Saturday=5, Sunday=6
# Check if the day of week is in [5, 6] (Saturday or Sunday)
df_flipkart['is_weekend'] = df_flipkart['order_date'].dt.dayofweek.isin([5, 6])

# Display sample orders showing weekday name and corresponding is_weekend flag
print('Flipkart Orders with engineered is_weekend feature (Sample View):')
weekend_view = df_flipkart[['order_id', 'order_date', 'weekday', 'is_weekend', 'item_name']]
print(weekend_view.head(8))

# Breakdown of orders placed on Weekend vs Weekdays
print('\nOrder Count by Weekend vs Weekday:')
print(df_flipkart['is_weekend'].value_counts().rename({True: 'Weekend', False: 'Weekday'}))
```

## 🎯 Key Takeaways & Best Practices
- **Format Specification**: Providing explicit format strings (e.g. `format='%Y-%m-%d %H:%M:%S'`) to `pd.to_datetime()` yields major performance speedups and eliminates ambiguous date parsing (e.g. Day-first vs Month-first).
- **Vectorized Accessor (`.dt`)**: Always leverage `.dt` for element-wise datetime attribute extraction (`.dt.year`, `.dt.month`, `.dt.day_name()`, `.dt.dayofweek`) rather than slower list comprehensions or lambda loops.
- **Frequency Resampling**: `df.resample()` functions like a temporal `groupby()`. Commonly used frequency aliases include `'D'` (daily), `'W'` (weekly), `'M'` / `'ME'` (month-end), and `'Q'` (quarterly).
- **Timezone Safety**: Distinguish between timezone-naive and timezone-aware timestamps. Use `.dt.tz_localize()` to attach a timezone to naive dates, and `.dt.tz_convert()` to shift between timezones.
- **Temporal Feature Engineering**: Features like `is_weekend`, `hour_of_day`, and `quarter` are critical predictive drivers for demand forecasting, sales estimation, and anomaly detection.
