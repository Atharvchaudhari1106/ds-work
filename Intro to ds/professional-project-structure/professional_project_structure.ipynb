## Task 5: Open-Source Project Structural Audit (`facebook/prophet`)
> **Problem Statement:** Clone a trending open-source data science project from GitHub (not the one built in class), inspect its folder structure, and list 3 ways it follows or differs from the recommended DA/DS structure. Suggest one improvement you would make to its organization.

### 1. Selected Project: **Meta's Prophet** (`https://github.com/facebook/prophet`)
**Cloning Command Executed:**
```bash
git clone --depth 1 https://github.com/facebook/prophet.git cloned-ds-project
```

**Top-Level Inspected Structure:**
```text
cloned-ds-project/
├── .github/          # CI/CD workflows & issue templates
├── docs/             # Sphinx/Docusaurus documentation source
├── examples/         # Quickstart example scripts
├── notebooks/        # 11 interactive diagnostic and demo notebooks
├── python/           # Python library core package, setup.py, and scripts
├── python_shim/      # Packaging backward compatibility
├── R/                # Complete R language implementation
├── docker-compose.yml
├── Dockerfile
└── README.md
```

### 2. Analysis: 3 Ways It Follows or Differs from the Recommended DA/DS Structure:

1. **Follows: Isolated `notebooks/` Directory for Demonstrations & Diagnostics:**  
   Like the canonical DA/DS pattern (Cookiecutter Data Science), Prophet segregates exploratory and diagnostic workflows (`quick_start.ipynb`, `diagnostics.ipynb`, `outliers.ipynb`, `handling_shocks.ipynb`) into a clean top-level `notebooks/` directory rather than cluttering the package root.

2. **Differs: Absence of an Explicit `data/raw/` vs. `data/processed/` Pipeline:**  
   Standard DA/DS projects maintain an immutable raw data store and a processed/cleaned feature store. In Prophet, benchmark datasets (such as the Peyton Manning Wikipedia pageviews time series) are embedded inside package test fixtures (`python/prophet/tests/`) or retrieved via URL inside notebooks, lacking an explicit `data/` data lifecycle hierarchy.

3. **Differs: Polyglot Dual-Language Layout (`python/` & `R/`) without `models/` Directory:**  
   Because Prophet is a distributed open-source modeling library rather than an applied end-to-end model pipeline, it splits into `python/` and `R/` directories, each with its own package manager files (`pyproject.toml` and R build configs). Furthermore, it does not maintain a `models/` or `saved_artifacts/` directory since model serialization occurs downstream in the end-user's application.

### 3. Suggested Structural Improvement:
- **Recommendation: Decouple sample datasets into a top-level `data/` directory with automated fetching scripts:**  
  Currently, sample CSV files used in demonstration notebooks are buried deep inside unit test fixtures. Creating a dedicated `data/` directory accompanied by a lightweight fetching script (`scripts/download_sample_datasets.py`) would adhere to the standard Data Science project structure and make running tutorials friction-free for newcomers.
