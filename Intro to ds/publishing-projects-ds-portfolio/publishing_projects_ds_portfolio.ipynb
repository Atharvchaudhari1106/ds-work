## 📦 Task 2: Git Large File Storage (Git LFS) & Data Engineering Rationale

### Why Raw Datasets (>100MB) Should NEVER Be Uploaded to Regular GitHub Repositories

1. **Repository Bloat & Immutable Git Object Store**:
   - Git is a Content-Addressable Storage (CAS) engine. Every revision of every file is stored as a compressed zlib object in `.git/objects/`.
   - Modifying a 100MB file 5 times bloats the `.git` directory by 500MB permanently. Even running `git rm` leaves the large binary in historical commits, forcing all collaborators and CI/CD pipelines to download gigabytes of dead data.
2. **GitHub Hard Limits & Bandwidth Saturation**:
   - GitHub blocks pushes containing individual files >100MB (`GH001: Large files detected`) and flags repositories >1GB.
   - Large raw files create massive network bottlenecks during automated CI/CD runs (e.g., GitHub Actions, Docker image building).
3. **Data Privacy, Governance & Regulatory Compliance (GDPR / HIPAA)**:
   - Raw datasets often contain Personally Identifiable Information (PII), unmasked IP addresses, or proprietary user event streams.
   - Under GDPR / CCPA, individuals possess the "Right to be Forgotten." Erasing a record from a raw CSV committed to Git requires rewriting cryptographic commit hashes across the entire repository history, disrupting all downstream clones and branches.
4. **Git LFS Architectural Solution**:
   - Git LFS replaces large binary blobs with lightweight 130-byte pointer files containing cryptographic SHA-256 hashes:
     ```text
     version https://git-lfs.github.com/spec/v1
     oid sha256:faf632dc82...
     size 111000570
     ```
   - The actual 105.86 MB payload is stored separately on dedicated object storage, preserving high-speed Git performance.


````python
# Verification of Git LFS Tracking and Sample Large Dataset Size
import os

lfs_file = os.path.join(repo_path, "data", "large_spotify_data.csv")
file_size_mb = os.path.getsize(lfs_file) / (1024 * 1024)

print(f"Sample Large Dataset: {lfs_file}")
print(f"File Size: {file_size_mb:.2f} MB (Meets >100MB constraint)")

print("\nGit LFS Tracking Attributes (.gitattributes):")
gitattrs_file = os.path.join(repo_path, '.gitattributes')
with open(gitattrs_file, 'r') as f:
    print(f.read().strip())

print("\nGit LFS Object Status:")
print(run_cmd(f'git -C "{repo_path}" lfs ls-files'))
````
