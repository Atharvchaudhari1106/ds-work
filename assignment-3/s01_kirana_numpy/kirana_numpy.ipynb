## Task 2: Store Tracker Arrays

**Objectives:**
Build the store's tracker arrays in the same notebook:
- `stock_levels` with `np.full((6, 7), 100)`
- `returns` with `np.zeros((6, 7))`
- `in_stock` with `np.ones(6, dtype=bool)`
- `item_ids` with `np.arange(101, 107)`
- `day_marks` with `np.linspace(0, 6, 7)`
- `item_pairs` with `np.eye(6)`
- Print each array's `dtype` and `nbytes`.
- *Hint applied:* Without `dtype=bool`, `np.ones` gives `1.0` values (float64) instead of `True`.


```python
# 1. stock_levels: 6 items x 7 days initialized to 100 units
stock_levels = np.full((6, 7), 100)

# 2. returns: 6 items x 7 days initialized to 0.0
returns = np.zeros((6, 7))

# 3. in_stock: boolean indicator for 6 items
in_stock = np.ones(6, dtype=bool)

# Demonstrate the hint: without dtype=bool, np.ones gives 1.0 (float64) instead of True
default_ones = np.ones(6)
print("Hint Demonstration:")
print("  np.ones(6) without dtype=bool -> dtype:", default_ones.dtype, "values:", default_ones)
print("  np.ones(6, dtype=bool)        -> dtype:", in_stock.dtype, "values:", in_stock)
print()

# 4. item_ids: SKU numbers 101 to 106
item_ids = np.arange(101, 107)

# 5. day_marks: 7 evenly spaced marks from day 0 to day 6
day_marks = np.linspace(0, 6, 7)

# 6. item_pairs: 6x6 identity matrix
item_pairs = np.eye(6)
```

```python
# Print each tracker array's dtype and nbytes
tracker_arrays = [
    ("stock_levels", stock_levels),
    ("returns", returns),
    ("in_stock", in_stock),
    ("item_ids", item_ids),
    ("day_marks", day_marks),
    ("item_pairs", item_pairs),
]

print("Tracker Arrays (dtype and nbytes):")
for name, arr in tracker_arrays:
    print(f"  {name:<13} -> dtype: {str(arr.dtype):<8} | nbytes: {arr.nbytes} bytes")
```

## Task 3: Python List vs. NumPy Array (Memory, Speed & Contiguity)

**Objectives:**
- Convert a Python list of 1,000,000 bill amounts (`list(range(1000000))`) into a NumPy array (`bills_arr`) and back with `.tolist()`.
- Print the list's `sys.getsizeof` next to the array's `nbytes`.
- *Hint applied:* `sys.getsizeof` on a list counts only its pointers, so label that number `'list pointers only'`.
- Print the `%timeit` result of `sum(bills_list)` next to `bills_arr.sum()`.
- Print the array's `flags['C_CONTIGUOUS']` and `strides`.


```python
# Create Python list of 1,000,000 bill amounts
bills_list = list(range(1000000))

# Convert to NumPy array
bills_arr = np.array(bills_list)

# Convert back to Python list with tolist()
bills_back = bills_arr.tolist()

print("Conversion Verification:")
print("  bills_list length :", len(bills_list))
print("  bills_arr length  :", len(bills_arr))
print("  bills_back length :", len(bills_back))
print("  Match verified    :", bills_list == bills_back)
```

```python
# Memory comparison: list sys.getsizeof vs array nbytes
list_pointers_size = sys.getsizeof(bills_list)
array_nbytes = bills_arr.nbytes

print("Memory Comparison:")
print(f"  sys.getsizeof (list pointers only): {list_pointers_size:,} bytes")
print(f"  bills_arr nbytes                  : {array_nbytes:,} bytes")
print()
# Context: In addition to the pointers, Python allocates a PyObject for each int (~28 bytes each)
total_list_estimate = list_pointers_size + (len(bills_list) * 28)
print(f"  Full Python list estimated memory : ~{total_list_estimate:,} bytes (~{total_list_estimate / (1024*1024):.2f} MB)")
print(f"  NumPy array total memory          : {array_nbytes:,} bytes ({array_nbytes / (1024*1024):.2f} MB)")
```

```python
# Speed comparison using %timeit
print("Timing sum(bills_list):")
%timeit sum(bills_list)

print()
print("Timing bills_arr.sum():")
%timeit bills_arr.sum()
```

```python
# Contiguous memory layout and strides
print("Memory Layout & Strides:")
print("  C_CONTIGUOUS :", bills_arr.flags['C_CONTIGUOUS'])
print("  strides      :", bills_arr.strides)
```

## Task 4: Array Report Function & Complete Summary

**Objectives:**
- Write a function `array_report(name, arr)` that prints one line with:
  - `name`
  - `shape`
  - `ndim`
  - `size`
  - `dtype`
  - `itemsize`
  - `nbytes`
- Call it for all nine arrays built:
  1. `units_sold`
  2. `price_inr`
  3. `stock_levels`
  4. `returns`
  5. `in_stock`
  6. `item_ids`
  7. `day_marks`
  8. `item_pairs`
  9. `bills_arr` (million-element array)


```python
def array_report(name, arr):
    """
    Prints a single line summarizing array attributes:
    name, shape, ndim, size, dtype, itemsize, and nbytes.
    """
    print(f"{name:<13} | shape: {str(arr.shape):<11} | ndim: {arr.ndim} | size: {arr.size:<7} | dtype: {str(arr.dtype):<8} | itemsize: {arr.itemsize}B | nbytes: {arr.nbytes:<8}B")
```

```python
# Call array_report for all nine arrays
all_nine_arrays = [
    ("units_sold", units_sold),
    ("price_inr", price_inr),
    ("stock_levels", stock_levels),
    ("returns", returns),
    ("in_stock", in_stock),
    ("item_ids", item_ids),
    ("day_marks", day_marks),
    ("item_pairs", item_pairs),
    ("bills_arr", bills_arr),
]

print("=" * 86)
print(f"{'Name':<13} | {'Shape':<18} | {'Ndim'} | {'Size':<12} | {'Dtype':<15} | {'Itemsize'} | {'Nbytes'}")
print("=" * 86)
for name, arr in all_nine_arrays:
    array_report(name, arr)
print("=" * 86)
```
