## Task 2: Non-Fast-Forward Merge, Conflict Generation & Resolution
> **Problem Statement:** Simulate a team workflow by merging your `feature-playlist` branch into `main` using a merge commit (not fast-forward), then intentionally create a merge conflict by editing the same line in `playlist.md` on both branches, and resolve the conflict using the VS Code merge tool.

### Step 2.1: Non-Fast-Forward Merge (`--no-ff`)
```bash
git checkout main
git merge --no-ff feature-playlist -m "Merge branch 'feature-playlist' into main (non-fast-forward)"
```

### Step 2.2: Intentionally Triggering a Merge Conflict
1. **On `main` Branch:** Edited Line 4 of `playlist.md` to:  
   `1. **Today's Top Hits:** Maintained by Main branch with dynamic daily viral chart updates.`  
   Committed on `main` (commit: `1bc24f0`).
2. **On `feature-playlist` Branch:** Edited the exact same Line 4 to:  
   `1. **Today's Top Hits:** Curated by Feature branch with weekly algorithmic recommendation filters.`  
   Committed on `feature-playlist` (commit: `4cc01cd`).
3. **Attempted Merge on `main`:**  
   ```bash
   git merge feature-playlist
   ```
   **Git Conflict Notice:**
   ```text
   Auto-merging playlist.md
   CONFLICT (content): Merge conflict in playlist.md
   Automatic merge failed; fix conflicts and then commit the result.
   ```

### Step 2.3: Observed Conflict Markers:
```text
<<<<<<< HEAD
1. **Today's Top Hits:** Maintained by Main branch with dynamic daily viral chart updates.
=======
1. **Today's Top Hits:** Curated by Feature branch with weekly algorithmic recommendation filters.
>>>>>>> feature-playlist
```

### Step 2.4: Conflict Resolution:
We synthesized both team modifications into a unified, superior description:
```markdown
1. **Today's Top Hits:** Synthesized hybrid feed featuring dynamic daily viral chart updates and weekly algorithmic recommendation filters.
```
Staged and committed the resolution:
```bash
git add playlist.md
git commit -m "Merge branch 'feature-playlist': resolve playlist.md description conflict"
```

## Task 4: Push to Fork & Open Pull Request (PR)
