### Overview: The Role of Git & GitHub in Data Science Workflows

Git is a distributed version control system (DVCS) that tracks changes in source code and notebook files over time, enabling reproducibility, seamless experimentation, and frictionless collaboration.

```
                LOCAL WORKSPACE                            REMOTE HOST (GITHUB)
  ┌───────────────────────────────────────────┐      ┌─────────────────────────────┐
  │ Working Directory ──▶ Staging Area (Index)│ ───▶ │ Remote Repository (origin)  │
  │   (README.md)           (git add)         │      │      (foodie-favorites)     │
  │        │                     │            │      │              ▲              │
  │        └─────────────────────┼────────────┘      └──────────────┼──────────────┘
  │                              ▼                                  │
  │                     Local Repository (.git) ────────────────────┘
  │                        (git commit)                   (git push)
  └───────────────────────────────────────────┘
```

## Task 5: Avoiding Notebook Versioning Chaos with Git & GitHub
> **Problem Statement:** Write a short note (3-4 lines) explaining how using Git and GitHub can help you avoid the problem of having multiple versions of the same Jupyter notebook file (like `analysis_v1.ipynb`, `analysis_final2.ipynb`, etc.) when working on a data science project.

---

### 💡 Short Note (3–4 Lines):

> *"Using Git provides a centralized, cryptographic commit history where each change is recorded as a clean snapshot under a single, canonical filename (`analysis.ipynb`), completely eliminating filename clutter like `analysis_v1.ipynb` or `analysis_final2.ipynb`.*  
> *Through isolated Git feature branches, data scientists can independently test speculative machine learning models or data cleaning transformations without risking corruption of the primary working codebase.*  
> *GitHub acts as an immutable cloud audit trail with visual cell-by-cell diffs, enabling teams to trace exact historical model performance, review pull requests collaboratively, and seamlessly roll back breaking bugs using a single command (`git revert` / `git checkout`)."*
