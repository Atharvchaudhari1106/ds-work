## Task 2: Create the `playlists` Table
> **Problem Statement:** Inside the `'music_streaming_app'` database, create a table called `'playlists'` with columns: `playlist_id` (integer, primary key), `name` (varchar), and `created_by` (varchar).

### Schema Specification

| Column Name | Data Type | Constraint | Description |
| :--- | :--- | :--- | :--- |
| `playlist_id` | `INT` / `INTEGER` | `PRIMARY KEY` | Unique identifier for each playlist |
| `name` | `VARCHAR(255)` | `NOT NULL` | The title or name of the playlist |
| `created_by` | `VARCHAR(100)` | `NOT NULL` | Username of the creator / curator |

### Standard SQL Syntax
```sql
CREATE TABLE playlists (
    playlist_id INT PRIMARY KEY,
    name VARCHAR(255) NOT NULL,
    created_by VARCHAR(100) NOT NULL
);
```


````python
# Drop existing table if it exists for a fresh run
cursor.execute("DROP TABLE IF EXISTS playlists;")

# Create playlists table
create_table_query = """
CREATE TABLE playlists (
    playlist_id INTEGER PRIMARY KEY,
    name VARCHAR(255) NOT NULL,
    created_by VARCHAR(100) NOT NULL
);
"""
cursor.execute(create_table_query)
conn.commit()

# Verify table schema from SQLite metadata
schema_info = cursor.execute("PRAGMA table_info(playlists);").fetchall()
df_schema = pd.DataFrame(schema_info, columns=["cid", "Column Name", "Type", "NotNull", "Default", "PK"])
print("[SUCCESS] Table 'playlists' created successfully. Schema:")
display(df_schema[["Column Name", "Type", "NotNull", "PK"]])
````

## Task 3: Insert Sample Rows
> **Problem Statement:** Insert three sample rows into the `'playlists'` table representing playlists like `'Bollywood Hits'`, `'Chill Vibes'`, and `'Workout Mix'`, each created by a different user.

### SQL Syntax
```sql
INSERT INTO playlists (playlist_id, name, created_by) VALUES
(1, 'Bollywood Hits', 'Amit'),
(2, 'Chill Vibes', 'Pooja'),
(3, 'Workout Mix', 'Rahul');
```


````python
# Sample playlist records created by different users
sample_playlists = [
    (1, 'Bollywood Hits', 'Amit'),
    (2, 'Chill Vibes', 'Pooja'),
    (3, 'Workout Mix', 'Rahul')
]

insert_query = """
INSERT INTO playlists (playlist_id, name, created_by)
VALUES (?, ?, ?);
"""
cursor.executemany(insert_query, sample_playlists)
conn.commit()

print(f"[SUCCESS] Inserted {len(sample_playlists)} sample playlists.")

# Retrieve and display all rows
df_all = pd.read_sql_query("SELECT * FROM playlists;", conn)
print("\nCurrent Contents of 'playlists' Table:")
display(df_all)
````

## Task 4: Filter Query (`WHERE created_by = 'Amit'`)
> **Problem Statement:** Write an SQL `SELECT` query to display all playlists created by the user `'Amit'` from the `'playlists'` table.  
> *Hint: Use the `WHERE` clause to filter by the `'created_by'` column.*

### SQL Query & Explanation
```sql
SELECT playlist_id, name, created_by
FROM playlists
WHERE created_by = 'Amit';
```
- **`SELECT`**: Specifies which columns to retrieve (or `*` for all columns).
- **`FROM playlists`**: Identifies the source table containing the records.
- **`WHERE created_by = 'Amit'`**: Applies a row-level filter predicate so that only records matching this condition are included in the result set.


````python
# Execute filtered query
query_amit = """
SELECT playlist_id, name, created_by
FROM playlists
WHERE created_by = 'Amit';
"""

df_amit = pd.read_sql_query(query_amit, conn)

print("Query Result: Playlists Created by 'Amit'")
display(df_amit)
````

### 🤖 AI Response (ChatGPT / Copilot)

In SQL databases, data is organized in a structured, relational grid. To understand the relationship between a **Table**, a **Row**, and a **Column**, let's take the real-world example of **Zomato**, India's popular food delivery app.

Imagine how Zomato tracks every meal order placed across millions of users daily.

---

### 1. The Table (`Orders`)
* **What it is:** A structured collection of related data stored in a tabular format (rows and columns). It represents an entire business entity.
* **Zomato Analogy:** The **entire "Orders" ledger**.
  * Just like Zomato has separate tables for `Users`, `Restaurants`, `Menu_Items`, and `Delivery_Partners`, the `Orders` table contains every transaction made on the platform.

```
+-----------------------------------------------------------------------------------------+
|                                    TABLE: Orders                                        |
+----------+------------------+---------------------+-------------------+-----------------+
| order_id | customer_name    | restaurant_name     | total_amount (Rs) | order_status    |
+----------+------------------+---------------------+-------------------+-----------------+
| 101      | Rohan Sharma     | Biryani Blues       | 450.00            | Delivered       |
| 102      | Priya Patel      | Domino's Pizza      | 620.00            | Out for Delivery|
| 103      | Amit Verma       | Punjab Grill        | 890.00            | Preparing       |
+----------+------------------+---------------------+-------------------+-----------------+
```

---

### 2. The Column (Attribute / Field)
* **What it is:** A vertical element that represents a specific attribute, characteristic, or property that every record in the table possesses. Each column enforces a specific data type (such as Integer, Varchar, Decimal, or Date).
* **Zomato Analogy:** An **individual detail or specification** of an order.
  * For example, the `restaurant_name` column tells you *where* the food is from.
  * The `total_amount` column tracks *how much* was charged.
  * The `order_status` column tracks *what state* the delivery is in (`'Preparing'`, `'Out for Delivery'`, `'Delivered'`).

---

### 3. The Row (Record / Tuple)
* **What it is:** A single horizontal entry in the table that groups together all the column values for one specific, individual instance of the entity.
* **Zomato Analogy:** **One specific order receipt placed by a hungry customer.**
  * For instance, looking across **Row 1**:
    `[101 | Rohan Sharma | Biryani Blues | 450.00 | Delivered]`
  * This row gives you the complete story of Rohan's specific lunch order. It connects the ID, customer name, restaurant, price, and status into a single coherent fact.

---

### 📊 Quick Comparison Summary

| Concept | SQL Terminology | Zomato Analogy | Real Example |
| :--- | :--- | :--- | :--- |
| **Table** | Relation / Entity | The entire order logbook | `Orders` table |
| **Column** | Attribute / Field | A specific feature of any order | `restaurant_name` (`VARCHAR`), `total_amount` (`DECIMAL`) |
| **Row** | Tuple / Record | One customer's specific order | `(101, 'Rohan Sharma', 'Biryani Blues', 450.00, 'Delivered')` |
