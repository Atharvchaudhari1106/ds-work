## Task 2: Music App Genre Data Classification & Reasoning
> **Problem Statement:** Given this sample data from a music app:
> `['Pop', 'Rock', 'Jazz', 'Hip-Hop', 'Pop', 'Rock', 'Jazz', 'Pop']`  
> classify the type of data and explain your reasoning in 2-3 lines.

### Statistical Classification:
The data is **Qualitative / Categorical Data**, specifically **Nominal Data**.

### Reasoning (2–3 Lines):
1. **Qualitative Labels:** The observations represent descriptive genre categories (text labels) rather than measurable numerical amounts.
2. **No Natural Order:** There is no intrinsic mathematical hierarchy, rank, or sequence among genres ('Pop' is not mathematically "greater than" or "less than" 'Rock').
3. **Permissible Operations:** We cannot perform arithmetic operations like addition or averaging; only frequency counts, proportions, and mode (most frequent genre) are meaningful.

````python
from collections import Counter
import matplotlib.pyplot as plt

music_sample = ['Pop', 'Rock', 'Jazz', 'Hip-Hop', 'Pop', 'Rock', 'Jazz', 'Pop']
genre_frequencies = Counter(music_sample)

print(f"Music Genre Dataset : {music_sample}")
print(f"Total Track Samples : {len(music_sample)}")
print(f"Unique Genres       : {list(genre_frequencies.keys())}")
print(f"Frequency Count     : {dict(genre_frequencies)}")
print(f"Mode (Most Popular) : '{genre_frequencies.most_common(1)[0][0]}' (count = {genre_frequencies.most_common(1)[0][1]})")
print(f"\nClassification Conclusion: CATEGORICAL (NOMINAL) DATA")

# Visualizing Nominal Category Frequencies
plt.figure(figsize=(8, 3.5))
genres = list(genre_frequencies.keys())
counts = list(genre_frequencies.values())
bars = plt.bar(genres, counts, color=['#3b82f6', '#10b981', '#f59e0b', '#8b5cf6'], width=0.5, edgecolor="black")
plt.title("Music App Genre Distribution (Frequency of Nominal Categories)", fontsize=11, fontweight="bold")
plt.xlabel("Genre Category", fontsize=10)
plt.ylabel("Play Frequency", fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.5)

for bar in bars:
    plt.text(bar.get_x() + bar.get_width()/2., bar.get_height() + 0.05, f"{int(bar.get_height())}", ha='center', va='bottom', fontweight='bold')

plt.show()
````

## Task 3: Improving App Features Using Statistics
> **Problem Statement:** Pick any one feature from an app you use (such as the number of likes on an Instagram post, or the rating of a restaurant on Zomato) and describe how statistics could help improve that feature for users.

### Selected Feature: **Zomato Restaurant Star Rating & Delivery Time Estimation**

Currently, many users notice that a brand-new restaurant with a single 5.0 rating from the owner outranks a beloved restaurant with 2,500 reviews averaging 4.8. Furthermore, delivery estimates often fail during rain or peak traffic.

Here is how statistics directly elevates this feature:

#### 1. Bayesian Weighted Rating (Shrinkage to the Mean)
Instead of a naive arithmetic mean $\frac{\sum \text{stars}}{n}$, statistics uses **Bayesian Weighted Averages** (similar to IMDB's Top 250 algorithm):
$$\text{Weighted Rating} = \left( \frac{v}{v + m} \right) R + \left( \frac{m}{v + m} \right) C$$
Where:
- $v$ = Number of reviews for the restaurant
- $m$ = Minimum threshold reviews needed (e.g., 50 reviews)
- $R$ = Arithmetic average rating of the restaurant
- $C$ = Mean rating across the entire city

*Benefit:* Prevents unproven 1-review restaurants from artificially dominating search results while rewarding consistently tested high-volume outlets.

#### 2. Percentile-Based Delivery Estimates (85th–90th Percentile vs. Mean)
Delivery times are heavily skewed by rush hour traffic and weather. A simple average (Mean = 30 mins) leads to customer disappointment if 20% of orders take 55 mins.
*Statistical Solution:* Displaying the **85th Percentile** duration guarantees that $85\%$ of deliveries arrive before the promised time, setting transparent expectations.

#### 3. Outlier and Fraud Review Detection
Competitor review-bombing (sudden spikes of 1-star ratings) is detected and filtered using **Z-Score and IQR anomalies** over a rolling 48-hour time window.

````python
# Demonstration of Bayesian Weighted Rating vs Naive Arithmetic Mean
def calculate_bayesian_rating(v, R, m=50, C=3.8):
    """
    v: number of reviews for the restaurant
    R: average rating of the restaurant
    m: minimum reviews threshold (prior weight)
    C: city-wide benchmark average
    """
    return (v / (v + m)) * R + (m / (v + m)) * C

restaurants = [
    {"Name": "Restaurant A (New, 1 fake review)", "Reviews": 1, "Raw Average": 5.0},
    {"Name": "Restaurant B (Famous, 2000 reviews)", "Reviews": 2000, "Raw Average": 4.8},
    {"Name": "Restaurant C (Local favorite, 120 reviews)", "Reviews": 120, "Raw Average": 4.5}
]

for rest in restaurants:
    rest["Bayesian Weighted Rating"] = round(calculate_bayesian_rating(rest["Reviews"], rest["Raw Average"]), 2)

df_ratings = pd.DataFrame(restaurants)
display(df_ratings)
````

## Task 4: Statistical Terms for a Spotify Playlist Recommendation Feature
> **Problem Statement:** Suppose you are building a new playlist recommendation feature for a Spotify-like app. List 3 basic statistical terms (such as mean, median, mode, variance, etc.) that would be useful in analyzing user listening data, and explain each with a one-line example.

Here are 3 fundamental statistical measures applied to music recommendation engines:

| # | Statistical Term | Role in Recommendation System | Concrete One-Line Example |
| :--- | :--- | :--- | :--- |
| **1** | **Mode** | Identifies the single most frequently consumed category (genre, artist, or language) to anchor user preference. | *Example:* A user's most frequently streamed genre is **"Acoustic Indie"** ($mode$), so the system generates a "Discover Acoustic Indie" daily mix. |
| **2** | **Median** | Measures the central tempo (Beats Per Minute / BPM) or song duration without being corrupted by extreme outliers (e.g., 10-hour ambient white noise tracks). | *Example:* The user's median listening tempo is **$120\text{ BPM}$**, so Spotify suggests upbeat tracks clustered around $115–125\text{ BPM}$ for workout sessions. |
| **3** | **Variance (or Standard Deviation)** | Measures the degree of musical diversity / variety in the user's listening habits. | *Example:* A user with **low variance** exclusively listens to Bollywood Lo-Fi, while a user with **high variance** jumps from Classical to Heavy Metal, prompting Spotify to recommend either highly focused or widely eclectic playlists. |

````python
import statistics

# User listening data demonstration for Recommendation System
user_tempos_bpm = [115, 118, 120, 122, 120, 125, 119, 60] # 60 is a rare ambient ballad outlier
user_genres = ['Acoustic Indie', 'Acoustic Indie', 'Rock', 'Acoustic Indie', 'Pop', 'Acoustic Indie']

# 1. Mode: Identify primary taste cluster
preferred_genre = statistics.mode(user_genres)

# 2. Median: Robust central tempo (resistant to the 60 BPM ballad outlier)
typical_tempo = statistics.median(user_tempos_bpm)

# 3. Variance / Std Dev: Measure taste diversity
tempo_dispersion = statistics.stdev(user_tempos_bpm)

print("=" * 60)
print("SPOTIFY PLAYLIST RECOMMENDATION - STATISTICAL PROFILE")
print("=" * 60)
print(f"1. Mode (Anchor Genre)       : {preferred_genre} -> Recommendation: Curate 'Top {preferred_genre}' Mix")
print(f"2. Median (Target Tempo)     : {typical_tempo:.1f} BPM -> Recommendation: Filter tracks between {typical_tempo-5:.0f}-{typical_tempo+5:.0f} BPM")
print(f"3. Std Deviation (Diversity) : {tempo_dispersion:.2f} BPM -> Recommendation: {'Eclectic Playlist' if tempo_dispersion > 15 else 'Focused Consistency Playlist'}")
print("=" * 60)
````
