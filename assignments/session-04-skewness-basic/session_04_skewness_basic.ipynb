### Overview: Understanding Skewness

**Skewness** measures the asymmetry of the probability distribution of a real-valued random variable about its mean.

| Distribution Type | Visual Shape | Mean vs. Median vs. Mode | Skewness ($S$) |
| :--- | :--- | :--- | :--- |
| **Left-Skewed (Negative)** | Long tail extends to the **left** | $\text{Mean} < \text{Median} < \text{Mode}$ | $S < 0$ |
| **Symmetrical (Normal)** | Bell-shaped, balanced halves | $\text{Mean} \approx \text{Median} \approx \text{Mode}$ | $S \approx 0$ |
| **Right-Skewed (Positive)** | Long tail extends to the **right** | $\text{Mean} > \text{Median} > \text{Mode}$ | $S > 0$ |

## Task 1: Swiggy Delivery Partner Daily Orders - Shape & Skewness
> **Problem Statement:** Given the daily order counts for a Swiggy delivery partner over 10 days:
> `[12, 15, 17, 14, 13, 16, 200, 18, 14, 15]`  
> plot a simple graph (hand-drawn or using any tool) and describe whether the data is left skewed, right skewed, or symmetrical. Explain your reasoning based on the shape.

```python
import statistics
import pandas as pd
import matplotlib.pyplot as plt

# Given daily order counts over 10 days
order_counts = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]
n = len(order_counts)

mean_val = statistics.mean(order_counts)
median_val = statistics.median(order_counts)
mode_val = statistics.mode(order_counts)
skew_val = pd.Series(order_counts).skew()

print(f"Daily Swiggy Order Counts : {order_counts}")
print(f"Sorted Order Counts       : {sorted(order_counts)}")
print(f"Number of Days (n)        : {n}")
print(f"--------------------------------------------------")
print(f"Mean Daily Orders         : {mean_val:.2f} orders")
print(f"Median Daily Orders       : {median_val:.1f} orders")
print(f"Mode Daily Orders         : {mode_val} orders")
print(f"Sample Skewness           : {skew_val:.2f} (Strong Positive Skew)")
print(f"==================================================")

# Plot distribution histogram
plt.figure(figsize=(9, 4.5))
bins = [10, 20, 50, 100, 150, 210]
plt.hist(order_counts, bins=bins, color='#3b82f6', edgecolor='black', alpha=0.75, label='Days Count')
plt.axvline(mean_val, color='#ef4444', linestyle='--', linewidth=2.5, label=f"Mean = {mean_val:.1f}")
plt.axvline(median_val, color='#10b981', linestyle='-', linewidth=2.5, label=f"Median = {median_val:.1f}")
plt.title("Distribution of Swiggy Daily Orders (Severe Right Skew)", fontsize=12, fontweight="bold")
plt.xlabel("Order Count per Day", fontsize=11)
plt.ylabel("Number of Days", fontsize=11)
plt.annotate(f"Extreme Outlier (200 orders)\npulling tail to the right",
             xy=(200, 1), xytext=(120, 4),
             arrowprops=dict(facecolor='#ef4444', shrink=0.08, width=1.5, headwidth=6),
             fontsize=10, fontweight="bold", color="#b91c1c")
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### Conclusion & Reasoning for Task 1:

1. **Distribution Type:**  
   The data is **Right-Skewed (Positively Skewed)**.

2. **Reasoning Based on the Shape:**  
   - **Tail Extends to the Right:** For 9 out of 10 days, order counts are clustered on the left between 12 and 18 orders. The single extraordinary day with **200 orders** stretches a long tail far out to the right.
   - **Mean > Median Relationship:** Because the mean sums all numbers, the extreme 200-order outlier inflates the **Mean to 33.40**, whereas the **Median remains grounded at 15.00**.
   - When $\text{Mean} (33.4) > \text{Median} (15.0)$, the distribution is right-skewed.

## Task 2: Instagram Influencer Followers - Mean vs. Median & Skewness
> **Problem Statement:** Take the following list of Instagram influencer follower counts (in thousands):
> `[5, 7, 8, 8, 9, 10, 12, 15, 95]`  
> Calculate the mean and median, then state which is higher and what that tells you about the skewness of the data.  
> *Hint:* If the mean is pulled away from the median, the data is skewed in that direction.

```python
import statistics

# Instagram influencer follower counts in thousands
followers = [5, 7, 8, 8, 9, 10, 12, 15, 95]
n = len(followers)

mean_val = sum(followers) / n
median_val = statistics.median(followers)
diff = mean_val - median_val

print(f"Follower Counts (in thousands) : {followers}")
print(f"Total Influencers Sampled (n)  : {n}")
print(f"Sum of Followers               : {sum(followers)}k")
print(f"--------------------------------------------------")
print(f"Mean Followers                 : {mean_val:.2f}k ({mean_val*1000:,.0f} followers)")
print(f"Median Followers               : {median_val:.1f}k ({median_val*1000:,.0f} followers)")
print(f"Difference (Mean - Median)     : +{diff:.2f}k")
print(f"==================================================")
```

### Findings & Skewness Interpretation for Task 2:

1. **Which is higher?**  
   The **Mean ($18.78\text{k}$)** is significantly higher than the **Median ($9.00\text{k}$)** by over $9,780$ followers.

2. **What this tells us about Skewness:**  
   - As noted in the hint, when the mean is pulled away from the median toward the right, the data is **Right-Skewed (Positively Skewed)**.
   - The single celebrity/mega-influencer with **$95\text{k}$ followers** is an extreme positive outlier that drastically drags the mean upward, while the median of **$9\text{k}$** remains representative of typical micro-influencers.

## Task 3: Real-World Example of Right-Skewed Data from Tech Apps
> **Problem Statement:** Find a real-world example of right-skewed data from any app you use (like Zomato order amounts, Paytm wallet balances, or YouTube video views). Briefly describe the dataset and explain why you think it is right skewed.

### Selected Example: **Zomato Food Delivery Order Bill Amounts**

#### 1. Dataset Description:
Consider the dataset of individual order checkout totals (in INR) across all users on Zomato in an Indian metropolitan city over a weekend.

#### 2. Why this data is Right-Skewed:
1. **Natural Lower Bound (Zero / Minimum Order Value):** Food order bills cannot be negative. Most individual orders cluster in a compact range between **₹150 and ₹450** (single meals, fast food, snacks).
2. **Open-Ended Upper Tail:** While typical users spend under ₹500, a small fraction of orders are placed for large family reunions, corporate team dinners, or luxury fine-dining that reach **₹5,000, ₹15,000, or even ₹30,000+**.
3. **Asymmetric Tail:** The huge bulk of orders create a towering peak on the left side of the graph, while the occasional massive bulk orders stretch out a long, thin tail to the right.
4. **Impact on Central Tendency:** The median order bill might be ₹320, but the arithmetic mean is pulled up to ₹580+ by high-value group orders. Hence, it is textbook **right-skewed**.

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulation of 1,000 Zomato order bill amounts (INR)
np.random.seed(42)
zomato_bills = np.random.lognormal(mean=5.8, sigma=0.6, size=1000)

mean_bill = np.mean(zomato_bills)
median_bill = np.median(zomato_bills)

print(f"Typical Median Order : INR {median_bill:.2f}")
print(f"Arithmetic Mean Bill : INR {mean_bill:.2f}")
print(f"Mean > Median confirms Right Skewness: {mean_bill > median_bill}")

# Visualizing the Right-Skewed distribution
plt.figure(figsize=(9, 4.2))
plt.hist(zomato_bills, bins=40, color='#f97316', edgecolor='black', alpha=0.75)
plt.axvline(mean_bill, color='#ef4444', linestyle='--', linewidth=2.5, label=f"Mean: INR {mean_bill:.0f}")
plt.axvline(median_bill, color='#10b981', linestyle='-', linewidth=2.5, label=f"Median: INR {median_bill:.0f}")
plt.title("Simulated Distribution of Zomato Order Values (Right-Skewed)", fontsize=12, fontweight="bold")
plt.xlabel("Order Bill Amount (INR)", fontsize=11)
plt.ylabel("Number of Orders", fontsize=11)
plt.xlim(0, 3000)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.show()
```

## Task 4: Flipkart Product Ratings Skewness
> **Problem Statement:** Given this dataset of Flipkart product ratings:
> `[3, 3, 4, 4, 4, 5, 5, 5, 5, 5]`  
> draw a rough graph and state if the distribution is symmetrical, left skewed, or right skewed. Justify your answer in one line.

```python
from collections import Counter
import statistics
import pandas as pd
import matplotlib.pyplot as plt

# Given Flipkart product ratings dataset
ratings = [3, 3, 4, 4, 4, 5, 5, 5, 5, 5]
counts = Counter(ratings)

mean_r = sum(ratings) / len(ratings)
med_r = statistics.median(ratings)
mode_r = statistics.mode(ratings)
skew_r = pd.Series(ratings).skew()

print(f"Flipkart Ratings Dataset : {ratings}")
print(f"Rating Counts            : {dict(counts)}")
print(f"--------------------------------------------------")
print(f"Mean Rating              : {mean_r:.2f}")
print(f"Median Rating            : {med_r:.2f}")
print(f"Mode Rating              : {mode_r}")
print(f"Skewness Coefficient     : {skew_r:.2f} (Negative)")
print(f"Inequality Check         : Mean ({mean_r:.2f}) < Median ({med_r:.2f}) < Mode ({mode_r})")
print(f"==================================================")

# Visual graph of ratings frequency
plt.figure(figsize=(8, 3.8))
stars = [1, 2, 3, 4, 5]
counts_stars = [counts.get(s, 0) for s in stars]
bars = plt.bar(stars, counts_stars, color=['#e2e8f0', '#e2e8f0', '#f59e0b', '#3b82f6', '#10b981'], width=0.55, edgecolor='black')
plt.axvline(mean_r, color='#ef4444', linestyle='--', linewidth=2.5, label=f"Mean = {mean_r:.2f}")
plt.axvline(med_r, color='#6366f1', linestyle='-', linewidth=2.5, label=f"Median = {med_r:.2f}")
plt.title("Flipkart Product Ratings Distribution (Left-Skewed)", fontsize=12, fontweight="bold")
plt.xlabel("Product Star Rating", fontsize=11)
plt.ylabel("Number of Reviews", fontsize=11)
plt.xticks(stars)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)

for bar in bars:
    if bar.get_height() > 0:
        plt.text(bar.get_x() + bar.get_width()/2., bar.get_height() + 0.1, f"{int(bar.get_height())}", ha='center', va='bottom', fontweight='bold')

plt.show()
```

### Answer & One-Line Justification for Task 4:

**Classification:**  
The distribution is **Left-Skewed (Negatively Skewed)**.

**One-Line Justification:**  
> *"The distribution is left-skewed because the ratings peak heavily on the high end at 5 stars while the tail trails off toward the lower 3-star ratings, resulting in $\text{Mean} (4.30) < \text{Median} (4.50) < \text{Mode} (5.00)$."*
