### Overview: Pearson's Correlation Coefficient ($r$)

**Correlation** measures the strength and direction of the linear relationship between two quantitative variables.

$$r = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum (x_i - \bar{x})^2 \sum (y_i - \bar{y})^2}}$$

| Range of $r$ | Interpretation | Direction |
| :--- | :--- | :--- |
| **$+0.7 \le r \le +1.0$** | Strong Positive Linear Association | As $X$ increases, $Y$ increases |
| **$+0.3 \le r < +0.7$** | Moderate Positive Association | Generally increases together |
| **$-0.3 < r < +0.3$** | Weak or Zero Linear Association | No clear linear trend |
| **$-0.7 < r \le -0.3$** | Moderate Negative Association | Generally moves in opposite directions |
| **$-1.0 \le r \le -0.7$** | Strong Negative Linear Association | As $X$ increases, $Y$ decreases |

## Task 1: Instagram Usage vs. Number of Posts (Scatter Plot & Visual Analysis)
> **Problem Statement:** Create two Python lists: one representing the number of hours you spend on Instagram each day for a week, and another for the number of posts you make each day. Plot these as a scatter plot using matplotlib and visually describe whether the relationship looks like a positive, negative, or zero correlation.

```python
import numpy as np
import matplotlib.pyplot as plt

# Two lists for 7 days (one week)
daily_hours_insta = [1.5, 2.0, 2.8, 3.5, 4.0, 4.8, 5.5]
daily_posts_insta = [1, 1, 2, 2, 3, 4, 4]

# Calculate correlation
r_insta = np.corrcoef(daily_hours_insta, daily_posts_insta)[0, 1]

print(f"7-Day Instagram Daily Hours (X) : {daily_hours_insta}")
print(f"7-Day Instagram Posts Made  (Y) : {daily_posts_insta}")
print(f"Calculated Pearson r            : {r_insta:.4f}")
print(f"Visual Classification           : Strong Positive Correlation")

# Scatter plot with trendline
plt.figure(figsize=(8, 4.2))
plt.scatter(daily_hours_insta, daily_posts_insta, color='#ec4899', s=90, edgecolors='black', zorder=3, label='Daily Log')

m, b = np.polyfit(daily_hours_insta, daily_posts_insta, 1)
x_vals = np.linspace(min(daily_hours_insta), max(daily_hours_insta), 50)
plt.plot(x_vals, m * x_vals + b, color='#475569', linestyle='--', linewidth=2, label=f'Trendline (r = +{r_insta:.2f})')

plt.title("Daily Instagram Hours vs. Posts Made (Positive Correlation)", fontsize=12, fontweight="bold")
plt.xlabel("Daily Time Spent on Instagram (Hours)", fontsize=11)
plt.ylabel("Number of Posts Created", fontsize=11)
plt.yticks([0, 1, 2, 3, 4, 5])
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Visual Description & Interpretation for Task 1:
- **Shape of the Points:** The scatter plot points ascend consistently from the bottom-left to the top-right.
- **Nature of Relationship:** The plot exhibits a **strong positive correlation ($r = +0.98$)**.
- **Substantive Meaning:** On days when more time is spent on the platform creating content, engagement is higher, leading to a higher number of posts published.

## Task 2: Correlation Between Zomato & Swiggy Monthly Orders
> **Problem Statement:** Given the following data for 6 friends: their monthly Zomato order count and their monthly Swiggy order count, calculate the correlation coefficient between the two lists using numpy's `corrcoef()`. Print the result and explain if the correlation is positive, negative, or zero.

We analyze the monthly order counts of 6 friends across both competing platforms.

```python
import numpy as np
import pandas as pd

# Data for 6 friends: monthly order counts
friends = ["Aarav", "Bhavya", "Chirag", "Divya", "Eshan", "Farhan"]
zomato_orders = [18, 12, 25, 6, 22, 10]
swiggy_orders = [15, 10, 22, 8, 20, 9]

# Display data table
df_food = pd.DataFrame({
    "Friend": friends,
    "Monthly Zomato Orders": zomato_orders,
    "Monthly Swiggy Orders": swiggy_orders
})
display(df_food)

# Calculate correlation matrix with numpy corrcoef
corr_matrix = np.corrcoef(zomato_orders, swiggy_orders)
r_val = corr_matrix[0, 1]

print("=" * 60)
print("CORRELATION ANALYSIS: ZOMATO VS SWIGGY ORDERS")
print("=" * 60)
print(f"Numpy Correlation Matrix:\n{corr_matrix}")
print(f"\nPearson Correlation Coefficient (r): {r_val:.4f}")
print("=" * 60)
```

### Explanation for Task 2:
1. **Result:** The correlation coefficient is **$r = +0.9859$** (near $+1.0$).
2. **Classification:** This represents an exceptionally **Strong Positive Correlation**.
3. **Real-World Explanation:**
   - Friends who frequently order online food (foodies/hostellers like Chirag with 25 Zomato orders) also place a high volume of orders on Swiggy (22 orders) to compare restaurant discounts and delivery times.
   - Conversely, friends who cook at home (like Divya with only 6 Zomato orders) also place very few orders on Swiggy (8 orders).
   - Thus, overall online food delivery appetite drives both variables upward in unison.

## Task 4: Negative Correlation Dataset & Scatter Plot (Spotify vs. Test Marks)
> **Problem Statement:** Use ChatGPT to generate a small dataset (10 pairs of numbers) that shows a strong negative correlation between time spent on Spotify and marks scored in a test. Paste the dataset and plot it as a scatter plot using matplotlib.  
> *Hint:* Ask ChatGPT to "generate 10 pairs of numbers with a strong negative correlation" and specify your variables.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 10 Pairs of data showing strong negative correlation
spotify_hours = [1.0, 1.5, 2.2, 3.0, 3.8, 4.5, 5.0, 5.8, 6.5, 7.2]
test_marks = [94, 89, 85, 78, 74, 68, 62, 55, 48, 42]

# Display dataset
df_spotify = pd.DataFrame({
    "Student": [f"Student {i+1}" for i in range(10)],
    "Spotify Hours (X)": spotify_hours,
    "Test Marks (Y)": test_marks
})
display(df_spotify)

# Compute correlation coefficient
r_neg = np.corrcoef(spotify_hours, test_marks)[0, 1]
print(f"Calculated Pearson Correlation (r): {r_neg:.4f}")

# Scatter plot with downward regression line
plt.figure(figsize=(9, 4.5))
plt.scatter(spotify_hours, test_marks, color='#1db954', s=100, edgecolors='black', zorder=3, label='Student Performance')

m, b = np.polyfit(spotify_hours, test_marks, 1)
x_line = np.linspace(min(spotify_hours), max(spotify_hours), 50)
plt.plot(x_line, m * x_line + b, color='#dc2626', linestyle='--', linewidth=2.5, label=f'Negative Fit (r = {r_neg:.3f})')

plt.title("Time Spent on Spotify vs. Test Marks (Strong Negative Correlation)", fontsize=12, fontweight="bold")
plt.xlabel("Daily Time Spent on Spotify (Hours)", fontsize=11)
plt.ylabel("Test Marks (Out of 100)", fontsize=11)
plt.ylim(30, 105)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
