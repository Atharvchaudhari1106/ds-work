### Overview: Foundations of Linear Algebra in Data Science

Linear algebra is the foundational mathematical language of modern Data Science and Machine Learning. Datasets in tabular formats, image pixels, feature embeddings, weights in neural networks, and recommendation matrices are all represented as **vectors** and **matrices**.

#### 1. Vectors (1D Arrays)
A vector is an ordered collection of scalar numbers representing a point or feature state in $n$-dimensional space:
$$\mathbf{v} = \begin{bmatrix} v_1 \\ v_2 \\ \vdots \\ v_n \end{bmatrix} \in \mathbb{R}^n$$
- In Python, vectors are modeled using **Python lists** `[x1, x2, ...]` or 1D **NumPy arrays** `np.array([x1, x2, ...])`.

#### 2. Matrices (2D Arrays)
A matrix is a two-dimensional grid of numbers arranged in $m$ rows and $n$ columns:
$$M = \begin{bmatrix} 
m_{11} & m_{12} & \dots & m_{1n} \\
m_{21} & m_{22} & \dots & m_{2n} \\
\vdots & \vdots & \ddots & \vdots \\
m_{m1} & m_{m2} & \dots & m_{mn}
\end{bmatrix} \in \mathbb{R}^{m \times n}$$
- In Data Science:
  - **Rows** typically represent individual samples, instances, observations, or time points (e.g., users, days, products).
  - **Columns** represent variables, attributes, or features (e.g., ratings, prices, quantities).

## Task 1: 1D Vector - Playlist Song Durations
> **Problem Statement:** Create a Python list to represent a playlist of 5 song durations in seconds (e.g., `[210, 180, 245, 200, 175]`). Print the total duration of the playlist.


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Representing 5 song durations in seconds as a Python list (1D Vector)
song_durations_sec = [210, 180, 245, 200, 175]
song_titles = ["Song 1 (Intro)", "Song 2 (Acoustic)", "Song 3 (Ballad)", "Song 4 (Upbeat)", "Song 5 (Outro)"]

# Vector representation in NumPy
v_playlist = np.array(song_durations_sec)

# 2. Calculating total playlist duration
total_seconds = sum(song_durations_sec)  # Native Python sum
total_seconds_np = np.sum(v_playlist)   # NumPy vector reduction

# Convert total duration to minutes and seconds
total_minutes = total_seconds // 60
remaining_seconds = total_seconds % 60

# Summary statistics
avg_duration = np.mean(v_playlist)
min_duration = np.min(v_playlist)
max_duration = np.max(v_playlist)

print("=" * 65)
print("TASK 1: SPOTIFY PLAYLIST DURATION (1D VECTOR ANALYSIS)")
print("=" * 65)
print(f"Playlist Vector v (seconds) : {song_durations_sec}")
print(f"Vector Dimensions           : 1D Vector of length {len(song_durations_sec)} (v in R^5)")
print("-" * 65)
print(f"TOTAL PLAYLIST DURATION     : {total_seconds} seconds")
print(f"FORMATTED DURATION          : {total_minutes} minutes and {remaining_seconds} seconds ({total_seconds/60:.2f} mins)")
print(f"Average Song Duration       : {avg_duration:.1f} seconds ({avg_duration/60:.2f} mins)")
print(f"Shortest Track Duration     : {min_duration} seconds")
print(f"Longest Track Duration      : {max_duration} seconds")
print("=" * 65)

# Visualizing the 1D Duration Vector
fig, ax = plt.subplots(figsize=(9, 4.5))

bars = ax.bar(song_titles, song_durations_sec, color='#1DB954', edgecolor='black', width=0.55, linewidth=1.2)
ax.axhline(avg_duration, color='#e74c3c', linestyle='--', linewidth=1.8, 
           label=f'Average Track Duration ({avg_duration:.1f}s)')

ax.set_ylabel('Duration (Seconds)', fontsize=11, fontweight='bold')
ax.set_title('Playlist Song Durations (1D Feature Vector $\\mathbf{v} \\in \\mathbb{R}^5$)', fontsize=13, fontweight='bold')
ax.set_ylim(0, 280)
ax.grid(axis='y', linestyle='--', alpha=0.5)
ax.legend(loc='upper right', frameon=True)

# Add value labels on bars
for bar in bars:
    h = bar.get_height()
    mins = h // 60
    secs = h % 60
    ax.annotate(f'{h}s\n({mins}m {secs}s)',
                xy=(bar.get_x() + bar.get_width() / 2, h),
                xytext=(0, 4), textcoords='offset points',
                ha='center', va='bottom', fontsize=9.5, fontweight='bold')

plt.tight_layout()
plt.show()

```

## Task 2: 3x3 Matrix - Swiggy Food Ratings by Friends
> **Problem Statement:** Represent a 3x3 matrix in Python to store the ratings (1-5) given by 3 friends to 3 different food items on Swiggy. Print the matrix and display the rating each friend gave to the second food item.

---

### Mathematical Representation:
Let the rating matrix $R \in \mathbb{R}^{3 \times 3}$ be defined as:

$$R = \begin{bmatrix}
r_{11} & r_{12} & r_{13} \\
r_{21} & r_{22} & r_{23} \\
r_{31} & r_{32} & r_{33}
\end{bmatrix}
=
\begin{matrix}
\text{Friend 1 (Aman)} \\
\text{Friend 2 (Priya)} \\
\text{Friend 3 (Rohan)}
\end{matrix}
\begin{bmatrix}
5 & 4 & 3 \\
4 & 5 & 2 \\

\end{bmatrix}
\begin{matrix}

$$

* **Rows (Index $i$):** 3 Friends $\to$ Aman ($i=0$), Priya ($i=1$), Rohan ($i=2$).
* **Columns (Index $j$):** 3 Food Items $\to$ Paneer Butter Masala ($j=0$), Butter Naan ($j=1$), Gulab Jamun ($j=2$).

## Task 3: Dish Orders Across Days (Matrix Column-Wise Reduction)
> **Problem Statement:** Given a matrix representing the number of orders for 4 different dishes across 3 days (rows: days, columns: dishes), write code to calculate and print the total orders for each dish.

---

### Mathematical Formalism:
Let matrix $M \in \mathbb{R}^{3 \times 4}$ represent the daily orders:

$$M = \begin{bmatrix}
m_{11} & m_{12} & m_{13} & m_{14} \\
m_{21} & m_{22} & m_{23} & m_{24} \\
m_{31} & m_{32} & m_{33} & m_{34}
\end{bmatrix}
=
\begin{matrix}
\text{Day 1 (Friday)} \\
\text{Day 2 (Saturday)} \\
\text{Day 3 (Sunday)}
\end{matrix}
\begin{bmatrix}
24 & 35 & 18 & 42 \\
30 & 48 & 25 & 55 \\
28 & 42 & 22 & 49
\end{bmatrix}$$

* **Total Orders per Dish:** Computed by summing along the column axis ($i=1$ to $3$ for each column $j$):
$$\mathbf{s} = \sum_{i=1}^{3} M_{i, j} = \mathbf{1}_{3}^T M = \begin{bmatrix} 1 & 1 & 1 \end{bmatrix} \begin{bmatrix} 
24 & 35 & 18 & 42 \\
30 & 48 & 25 & 55 \\
28 & 42 & 22 & 49
\end{bmatrix} = \begin{bmatrix} 82 & 125 & 65 & 146 \end{bmatrix}$$

In NumPy, this corresponds to **Axis 0 reduction:** `np.sum(M, axis=0)`.



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Define order matrix (3 Days x 4 Dishes)
# Rows: Days (Day 1: Friday, Day 2: Saturday, Day 3: Sunday)
# Columns: Dishes (Paneer Biryani, Butter Chicken, Veg Manchurian, Masala Dosa)
days = ["Day 1 (Friday)", "Day 2 (Saturday)", "Day 3 (Sunday)"]
dishes = ["Paneer Biryani", "Butter Chicken", "Veg Manchurian", "Masala Dosa"]

orders_matrix = [
    [24, 35, 18, 42],  # Day 1 orders
    [30, 48, 25, 55],  # Day 2 orders
    [28, 42, 22, 49]   # Day 3 orders
]

M = np.array(orders_matrix)

# 2. Method A: Native Python Nested Loop / List Comprehension
num_dishes = len(dishes)
total_orders_python = [sum(orders_matrix[day][d] for day in range(len(days))) for d in range(num_dishes)]

# 3. Method B: NumPy Column-Wise Sum (axis=0)
total_orders_numpy = np.sum(M, axis=0)

# Build Comprehensive DataFrame for Display
df_orders = pd.DataFrame(M, index=days, columns=dishes)
df_orders.loc['TOTAL ORDERS'] = total_orders_numpy

print("=" * 70)
print("TASK 3: RESTAURANT ORDERS MATRIX (3 DAYS x 4 DISHES)")
print("=" * 70)
print("Orders Matrix (Shape: 3 Rows x 4 Columns):")
print(M)
print("\nTabular Matrix with Column Totals:")
print(df_orders.to_string())

print("-" * 70)
print("TOTAL ORDERS CALCULATED FOR EACH DISH:")
print("-" * 70)
for dish_name, total in zip(dishes, total_orders_numpy):
    print(f"  * {dish_name:<20} : {total:>4} total orders")

most_popular_dish = dishes[np.argmax(total_orders_numpy)]
max_orders = np.max(total_orders_numpy)
print("-" * 70)
print(f"Top Performing Dish: '{most_popular_dish}' with {max_orders} orders.")
print(f"Overall Orders Across All Dishes & Days: {np.sum(M)} orders.")
print("=" * 70)

# VISUALIZATION: Daily Breakdown & Total Orders per Dish
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Stacked Bar Chart of Daily Orders
x = np.arange(len(dishes))
width = 0.5
bottom = np.zeros(len(dishes))
day_colors = ['#3498db', '#e74c3c', '#f1c40f']

for i, day in enumerate(days):
    ax1.bar(x, M[i, :], width, bottom=bottom, label=day, color=day_colors[i], edgecolor='black')
    bottom += M[i, :]

ax1.set_ylabel('Number of Orders', fontsize=11, fontweight='bold')
ax1.set_title('Daily Orders Breakdown per Dish (Stacked Matrix View)', fontsize=12, fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels(dishes, fontsize=9.5, fontweight='bold', rotation=12, ha='right')
ax1.legend(loc='upper left')
ax1.grid(axis='y', linestyle='--', alpha=0.5)

# Annotate total on top of stacks
for i in range(len(dishes)):
    ax1.annotate(f'Total: {int(bottom[i])}',
                 xy=(x[i], bottom[i]),
                 xytext=(0, 4), textcoords='offset points',
                 ha='center', va='bottom', fontsize=10, fontweight='bold')

# Plot 2: Total Orders Ranking Bar Chart
dish_colors = ['#27ae60', '#e67e22', '#9b59b6', '#2980b9']
bars2 = ax2.bar(dishes, total_orders_numpy, color=dish_colors, edgecolor='black', width=0.5, linewidth=1.2)
ax2.set_ylabel('Total Orders Count', fontsize=11, fontweight='bold')
ax2.set_title('Total Orders per Dish ($\sum_{i=1}^3 M_{i,j}$ via Axis 0 Sum)', fontsize=12, fontweight='bold')
ax2.set_ylim(0, 175)
ax2.set_xticklabels(dishes, fontsize=9.5, fontweight='bold', rotation=12, ha='right')
ax2.grid(axis='y', linestyle='--', alpha=0.5)

for bar in bars2:
    h = bar.get_height()
    ax2.annotate(f'{h}',
                 xy=(bar.get_x() + bar.get_width() / 2, h),
                 xytext=(0, 4), textcoords='offset points',
                 ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()

```

## Task 4: Flipkart Products Matrix & Revenue Calculation
> **Problem Statement:** Imagine you have a table of 5 Flipkart products with columns for price and quantity sold. Represent this data as a matrix in Python, then write a function to calculate the total revenue for each product.  
> *Hint: Multiply price and quantity for each row, then print the results.*

---

### Mathematical & Algorithmic Formulation:
Let $P \in \mathbb{R}^{5 \times 2}$ be the Flipkart product table represented as a $5 \times 2$ matrix:

$$P = \begin{bmatrix}
\text{Price}_1 & \text{Quantity}_1 \\
\text{Price}_2 & \text{Quantity}_2 \\
\text{Price}_3 & \text{Quantity}_3 \\
\text{Price}_4 & \text{Quantity}_4 \\
\text{Price}_5 & \text{Quantity}_5
\end{bmatrix}
=
\begin{matrix}
\text{Wireless Earbuds} \\
\text{Smart Fitness Band} \\
\text{Ergonomic Laptop Stand} \\
\text{Fast USB-C Charger} \\
\text{Noise-Cancelling Headphones}
\end{matrix}
\begin{bmatrix}
1499 & 120 \\
2499 & 85 \\
899 & 210 \\
699 & 340 \\
4999 & 45
\end{bmatrix}$$

* **Row-Wise Product Revenue:** For each product $i \in \{1, \dots, 5\}$:
$$\text{Revenue}_i = P_{i, 0} \times P_{i, 1}$$
* **Hadamard (Element-Wise) Vector Product:**
$$\mathbf{r} = \mathbf{p} \odot \mathbf{q} = \begin{bmatrix} 1499 \\ 2499 \\ 899 \\ 699 \\ 4999 \end{bmatrix} \odot \begin{bmatrix} 120 \\ 85 \\ 210 \\ 340 \\ 45 \end{bmatrix} = \begin{bmatrix} 179880 \\ 212415 \\ 188790 \\ 237660 \\ 224955 \end{bmatrix}$$

* **Total Store Revenue (Vector Dot Product):**
$$\text{Total Store Revenue} = \sum_{i=1}^5 \text{Revenue}_i = \mathbf{p}^T \mathbf{q} = \text{₹} 1,043,700$$



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Product catalog definitions
product_names = [
    "Wireless Earbuds",
    "Smart Fitness Band",
    "Ergonomic Laptop Stand",
    "Fast USB-C Charger",
    "Noise-Cancelling Headphones"
]

# 2. Representing Flipkart table as a 5x2 Matrix
# Column 0: Price (in INR)
# Column 1: Quantity Sold (units)
flipkart_data = [
    [1499, 120],  # Earbuds: ₹1,499 x 120
    [2499, 85],   # Smart Band: ₹2,499 x 85
    [899, 210],   # Laptop Stand: ₹899 x 210
    [699, 340],   # USB-C Charger: ₹699 x 340
    [4999, 45]    # Headphones: ₹4,999 x 45
]

# Matrix P in NumPy
P = np.array(flipkart_data)


# 3. Defining Dedicated Function to Calculate Total Revenue for Each Product
def calculate_product_revenue(matrix, names=None):
    """
    Calculates total revenue for each product in a (N x 2) price-quantity matrix.
    
    Formula:
        Revenue_i = Price_i * Quantity_i (Row-wise multiplication)
        
    Parameters:
    -----------
    matrix : list or np.ndarray
        A 2D array of shape (N, 2), where column 0 is Price and column 1 is Quantity.
    names : list, optional
        List of product names corresponding to the rows.
        
    Returns:
    --------
    np.ndarray:
        A 1D vector containing the revenue for each product.
    """
    mat = np.array(matrix)
    if mat.ndim != 2 or mat.shape[1] < 2:
        raise ValueError("Input matrix must be a 2D array with at least 2 columns [Price, Quantity].")
        
    # Extract price column and quantity column
    prices = mat[:, 0]
    quantities = mat[:, 1]
    
    # Row-wise element-wise multiplication (Hadamard product)
    revenues = prices * quantities
    return revenues


# 4. Execute function
revenues = calculate_product_revenue(P, product_names)
total_store_revenue = np.sum(revenues)

# Create Comprehensive Reporting DataFrame
df_flipkart = pd.DataFrame({
    "Product Name": product_names,
    "Unit Price (INR)": [f"₹{p:,}" for p in P[:, 0]],
    "Quantity Sold": P[:, 1],
    "Total Revenue (INR)": [f"₹{r:,}" for r in revenues]
})

print("=" * 75)
print("TASK 4: FLIPKART PRODUCTS REVENUE CALCULATION (MATRIX OPERATIONS)")
print("=" * 75)
print("Raw Price-Quantity Matrix P (Shape 5x2):")
print(P)
print("\nProduct Sales and Revenue Breakdown Table:")
print(df_flipkart.to_string(index=False))

print("-" * 75)
print("ROW-WISE REVENUE RESULTS:")
print("-" * 75)
for i, name in enumerate(product_names):
    price = P[i, 0]
    qty = P[i, 1]
    rev = revenues[i]
    print(f"  * {name:<28} : ₹{price:>5} x {qty:>3} units = ₹{rev:>8,}")

print("-" * 75)
print(f"TOTAL FLIPKART STORE REVENUE (Dot Product p^T q) : ₹{total_store_revenue:,}")
print(f"Top Revenue Generating Product                  : '{product_names[np.argmax(revenues)]}' (₹{np.max(revenues):,})")
print(f"Highest Volume Product Sold                     : '{product_names[np.argmax(P[:, 1])]}' ({np.max(P[:, 1])} units)")
print("=" * 75)

# VISUALIZATION: Revenue per Product and Price vs Quantity Comparison
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.5))

# Plot 1: Horizontal Bar Chart of Revenue per Product
y_pos = np.arange(len(product_names))
colors = ['#2980b9', '#8e44ad', '#27ae60', '#e67e22', '#c0392b']

bars = ax1.barh(y_pos, revenues / 1000, color=colors, edgecolor='black', height=0.55, linewidth=1.2)
ax1.set_yticks(y_pos)
ax1.set_yticklabels(product_names, fontsize=10, fontweight='bold')
ax1.invert_yaxis()  # Top-down order
ax1.set_xlabel('Revenue (in Thousands INR, ₹000s)', fontsize=11, fontweight='bold')
ax1.set_title('Total Revenue Generated per Flipkart Product\n($\\text{Revenue} = \\text{Price} \\times \\text{Quantity}$)', 
              fontsize=12, fontweight='bold')
ax1.grid(axis='x', linestyle='--', alpha=0.5)

for bar in bars:
    w = bar.get_width()
    ax1.annotate(f' ₹{w:.1f}k',
                 xy=(w, bar.get_y() + bar.get_height() / 2),
                 xytext=(3, 0), textcoords='offset points',
                 ha='left', va='center', fontsize=10, fontweight='bold')

# Plot 2: Scatter plot of Unit Price vs Quantity with Revenue bubble size
scatter = ax2.scatter(P[:, 0], P[:, 1], s=revenues / 500, c=revenues, cmap='viridis', 
                      edgecolor='black', linewidth=1.5, alpha=0.85)

for i, txt in enumerate(product_names):
    ax2.annotate(txt, (P[i, 0], P[i, 1]), textcoords="offset points", xytext=(0, 10), 
                 ha='center', fontsize=8.5, fontweight='bold')

ax2.set_xlabel('Unit Price (₹)', fontsize=11, fontweight='bold')
ax2.set_ylabel('Quantity Sold (Units)', fontsize=11, fontweight='bold')
ax2.set_title('Price vs. Quantity Dynamics\n(Bubble Size Proportional to Revenue)', fontsize=12, fontweight='bold')
ax2.grid(True, linestyle='--', alpha=0.5)
cbar = fig.colorbar(scatter, ax=ax2)
cbar.set_label('Revenue (INR)', fontweight='bold')

plt.tight_layout()
plt.show()

```
